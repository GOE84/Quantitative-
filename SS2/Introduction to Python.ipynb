## คอมเมนต์ในโค้ด (Code Comments)
คอมเมนต์ (Comment) คือบันทึกที่โปรแกรมเมอร์เขียนไว้ในซอร์สโค้ดของโปรแกรม โดยมีจุดประสงค์เพื่ออธิบายซอร์สโค้ดให้ชัดเจนและช่วยให้ผู้อื่นติดตามทำความเข้าใจได้ง่ายขึ้นว่าโค้ดกำลังทำอะไรอยู่ ข้อความใดๆ ที่อยู่ในคอมเมนต์จะถูกข้ามไปเมื่อมีการรันโค้ดจริง ทำให้คอมเมนต์มีประโยชน์อย่างมากสำหรับการใส่คำอธิบาย เหตุผล รวมถึงใช้ปิดการทำงาน (comment out) โค้ดบางบรรทัดที่คุณยังไม่แน่ใจ คอมเมนต์ใน Python สร้างขึ้นโดยใช้เครื่องหมายสี่เหลี่ยมหรือแฮช (# ใส่ข้อความที่นี่) การใส่ `#` ลงในบรรทัดโค้ดจะทำให้ข้อความทั้งหมดที่ตามหลังเครื่องหมายนั้นกลายเป็นคอมเมนต์

```python
# this comment
# โค้ดบรรทัดเหล่านี้จะไม่เปลี่ยนแปลงค่าใดๆ
# อะไรก็ตามที่ตามหลังเครื่องหมาย # ตัวแรกจะไม่ถูกนำไปรันเป็นโค้ด
```

คุณอาจเคยได้ยินคนเรียกข้อความที่ครอบด้วยเครื่องหมายคำพูดสามตัว (`""" ใส่ข้อความที่นี่ """`) ว่าเป็นคอมเมนต์แบบหลายบรรทัด (multi-line comments) แต่จริงๆ แล้วไม่ถูกต้องทั้งหมด สิ่งนี้คือ `string` ชนิดพิเศษ (ชนิดข้อมูลที่เราจะเรียนกันต่อไป) ที่เรียกว่า `docstring` ซึ่งใช้สำหรับอธิบายวัตถุประสงค์ของฟังก์ชัน

```python
""" นี่คือสตริงชนิดพิเศษ """
```

อย่าลืมอ่านคอมเมนต์ภายในแต่ละ Code Cell (ถ้ามี)
เพราะคอมเมนต์เหล่านั้นจะช่วยอธิบายแบบเรียลไทม์ว่าเกิดอะไรขึ้นบ้างในแต่ละบรรทัดของโค้ด


### ตัวแปร (Variables)
ตัวแปร (Variables) ทำหน้าที่ตั้งชื่อให้กับค่าต่างๆ ในการเขียนโปรแกรม หากคุณต้องการบันทึกค่าไว้ใช้ภายหลังหรือใช้ซ้ำ คุณสามารถตั้งชื่อให้กับค่านั้นและเก็บข้อมูลไว้ในตัวแปรได้ ตัวแปรในการเขียนโปรแกรมทำงานคล้ายกับตัวแปรในวิชาพีชคณิตโดยพื้นฐาน แต่ใน `Python` ตัวแปรสามารถเก็บข้อมูลได้หลายชนิดแตกต่างกัน

ชนิดของตัวแปรพื้นฐานที่เราจะครอบคลุมในหัวข้อนี้ ได้แก่ integers (จำนวนเต็ม), `floating point numbers` (จำนวนทศนิยม), `booleans` (ค่าความจริง), และ`strings` (ข้อความ)

integer ในการเขียนโปรแกรมมีความหมายเหมือนกับในคณิตศาสตร์ คือจำนวนเต็มที่ไม่มีค่าหลังจุดทศนิยม ในที่นี้เราจะใช้ฟังก์ชัน `print` ที่มีมาให้ในภาษา (built-in) เพื่อแสดงค่าของตัวแปรรวมถึงชนิดของตัวแปรด้วย!



```python
integer = 50
print(integer ,type(integer))
```

ตัวแปรไม่ว่าจะเป็นชนิดใดก็ตาม จะถูกกำหนดค่าโดยใช้เครื่องหมายเท่ากับตัวเดียว `(=)` ชื่อตัวแปรใน Python เป็นแบบ case-sensitive (ตัวพิมพ์เล็ก-ใหญ่ถือเป็นคนละตัวกัน) ดังนั้นการเปลี่ยนตัวพิมพ์ใหญ่-เล็กในชื่อตัวแปรแม้เพียงเล็กน้อย จะหมายถึงตัวแปรคนละตัวกันโดยสิ้นเชิง

```python
one = 1
print(one)
```

จำนวน `floating point` หรือเรียกสั้นๆ ว่า `float` เป็นชื่อเรียกทางการของจำนวนจริง (เช่นเดียวกับในคณิตศาสตร์) ในการกำหนดค่า `float` เราจำเป็นต้องใส่จุดทศนิยม หรือระบุให้ชัดเจนว่าค่านั้นเป็น `float`

```python
my_float = 1.0
print(my_float, type(my_float))

my_float = float(1)
print(my_float, type(my_float))
```

ตัวแปรชนิด `float` จะไม่ปัดเศษตัวเลขที่คุณเก็บไว้ในตัวแปร ในขณะที่ตัวแปรชนิด `integer` จะปัดทิ้ง สิ่งนี้ทำให้ `float` เหมาะสมกับการคำนวณทางคณิตศาสตร์ที่ต้องการความละเอียดมากกว่าแค่จำนวนเต็มมากกว่า

สังเกตว่าเช่นเดียวกับที่เราใช้ฟังก์ชัน `float()` เพื่อบังคับให้ตัวเลขถูกมองว่าเป็น float เราก็สามารถใช้ฟังก์ชัน `int()` เพื่อบังคับให้ตัวเลขถูกมองว่าเป็น `int` ได้เช่นกัน

```python
my_int = int(3.14159)
print(my_int, type(my_int))
```

ฟังก์ชัน `int()` จะตัดตัวเลขทั้งหมดที่อยู่หลังจุดทศนิยมทิ้งด้วย!

ส่วน Strings (สตริง) ช่วยให้คุณสามารถนำข้อความมาเก็บเป็นตัวแปรเพื่อนำไปใช้งานได้ โดยสามารถกำหนดได้ด้วยเครื่องหมายคำพูดเดี่ยว `('')` หรือเครื่องหมายคำพูดคู่ `("")`



```python
my_string = 'This is a string with single quotes'
print(my_string)
my_string = 'This is a string with double quotes'
print(my_string)
```

สามารถใช้ได้ทั้งสองแบบ เพื่อที่เราจะได้สามารถใส่เครื่องหมายอะพอสทรอฟี `(')` หรือเครื่องหมายคำพูด `(")` ไว้ในสตริงได้ตามต้องการ
